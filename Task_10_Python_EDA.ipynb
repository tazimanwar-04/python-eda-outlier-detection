```python
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
print("Q1:", Q1)
print("Q3:", Q3)
```

```python
IQR = Q3 - Q1
print("IQR:", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[(df['price'] < lower_bound) | (df['price'] > upper_bound)]
print("Number of outliers:", outliers.shape[0])
```

```python
print("Max price before capping:", df['price'].max())
```

```python
df['price']= df['price'].clip(lower=lower_bound, upper= upper_bound)
```

```python
print("max price after capping:", df['price'].max())
```

```python
new_outliers = df[(df['price'] > upper_bound)]
print("Outliers after capping:", new_outliers.shape[0])
```

```python
df['price_outlier_flag'] = 0
df.loc[df['price']== upper_bound, 'price_outlier_flag'] = 1
```

```python
df['price_outlier_flag'].value_counts()
```

```python
correlation = df.corr(numeric_only=True)
correlation['price'].sort_values(ascending=False)
```

```python
df.to_csv("cleaned_dataset.csv", index=False)
```
